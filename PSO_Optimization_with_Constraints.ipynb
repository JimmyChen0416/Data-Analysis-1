```python
# First, you need to install pyswarm. Run this in a separate cell:
# !pip install pyswarm #Restart Kernel after installing pyswarm

import numpy as np #NumPy 是 Python 做：數值計算,矩陣運算,陣列處理,常用工具。
from pyswarm import pso  # This will work after installation


def objective(x): #建立一個函式
    x1, x2 = x
    return x1**4 - 2 * x2 * x1**2 + x2**2 + x1**2 - 2 * x1 + 5#目標公式


lb = [-3, -1]  # Lower bounds #x1最低為-3 #x2最低為-1
ub = [2, 6]    # Upper bounds #x1最高為2 #x2最高為6


# Run particle swarm optimization
xopt, fopt = pso(objective, lb, ub)#使用 PSO，在 lb 和 ub 限制範圍內，尋找讓 objective 最佳的答案。

print("Optimal solution:", xopt)#給出最佳參數
print("Function value:", fopt) #給出最佳函數值
```

```python
import numpy as np # 匯入 NumPy 套件，用於數值計算、陣列處理、矩陣運算
from pyswarm import pso
# 從 pyswarm 套件匯入 PSO 演算法函式
# pso() 是用來執行 Particle Swarm Optimization（粒子群最佳化）

# 目標函數
def objective(x):
    # 定義 PSO 要最佳化的目標函數
    # PSO 會不斷嘗試不同的 x 組合，並把結果回傳比較
    X = x[0]
    # 將 PSO 傳入的第一個參數設定為 X
    # 例如 x=[10,20,30]，則 X=10
    Y = x[1]
    # 將 PSO 傳入的第二個參數設定為 Y
    # 例如 x=[10,20,30]，則 Y=20
    Z = x[2]
    # 將 PSO 傳入的第三個參數設定為 Z
    # 例如 x=[10,20,30]，則 Z=30
    return  2*X-4*Y+5*Z+3*X*Y
     # 目標函數公式
    # 將 X、Y、Z 帶入公式計算結果
    # PSO 會根據這個結果判斷哪一組 X,Y,Z 比較好
    # pyswarm 預設是尋找最小值

# 限制條件
def constraint(x):
      # 定義限制條件，限制 PSO 搜尋的答案必須符合要求
    X = x[0]
     # 取得第一個變數 X
    Y = x[1]
    # 取得第二個變數 Y
    Z = x[2]
    # 取得第三個變數 Z
    return 300 - (X + 2*Y + 3*Z)
    # 限制條件：
    # X + 2Y + 3Z <= 300
    # pyswarm 判斷限制條件時，需要回傳 >=0 才代表符合
    # 所以改寫成：
    # 300 - (X+2Y+3Z) >= 0
    # 等同於：
    # X+2Y+3Z <=300

# 搜尋範圍
lb = [0,0,0]
# Lower Bound，下限
# X、Y、Z 最小值
# 代表：
# X >= 0
# Y >= 0
# Z >= 0

ub = [100,100,100]
# Upper Bound，上限
# X、Y、Z 最大值
# 代表：
# X <=100
# Y <=100
# Z <=100

# 執行PSO
xopt, fopt = pso( objective, lb,ub, f_ieqcons=constraint,maxiter=100 )
# objective:
# 告訴 PSO 要最佳化哪個公式
#
# lb:
# 告訴 PSO 每個變數的最低搜尋範圍
#
# ub:
# 告訴 PSO 每個變數的最高搜尋範圍
#
# f_ieqcons:
# 加入限制條件 constraint
# 確保找到的答案符合 X+2Y+3Z<=300
#
# maxiter=100:
# PSO 最大迭代次數
# 代表粒子更新100次後停止搜尋


print("最佳X,Y,Z:",xopt)
# 印出 PSO 找到的最佳參數組合
# xopt 例如：
# [20,30,10]
# 代表最佳 X=20,Y=30,Z=10

print("最佳目標值:",fopt)
# 印出最佳參數組合代入 objective 後得到的結果
# 也就是最佳化後的函數值

#先將 X、Y、Z 設定為決策變數，建立目標函數 objective，再透過 LB、UB 限制變數搜尋範圍
#加入 constraint 限制條件，最後利用 PSO 透過多次迭代搜尋最佳參數組合，並取得最佳目標函數值。
```

```python
#假如這次想手動輸入數值的話.....
import numpy as np
from pyswarm import pso

# 目標函數
def objective(x):
    X = x[0]
    Y = x[1]
    Z = x[2]
    value = 2*X - 1*Y + 5*Z + 3*X*Y

    return -value

# 限制條件
def constraint(x):
    X = x[0]
    Y = x[1]
    Z = x[2]
    return 300 - (X + 2*Y + 3*Z)

# PSO搜尋範圍
lb = [0,0,0]
ub = [100,100,100]


# 執行PSO
xopt, fopt = pso( objective, lb, ub, f_ieqcons=constraint, maxiter=100)


print("PSO最佳X,Y,Z:", xopt)
print("PSO最佳目標值:",-fopt)



# =========================
# 手動輸入測試
# =========================

while True:

    # 輸入X
    X = float(input("請輸入X值(0~100): "))

    # 輸入Y
    Y = float(input("請輸入Y值(0~100): "))

    # 輸入Z
    Z = float(input("請輸入Z值(0~100): "))


    # 檢查 X,Y,Z 範圍
    if X > 100 or X < 0:
        print("X範圍只能0~100，請重新輸入")

    elif Y > 100 or Y < 0:
        print("Y範圍只能0~100，請重新輸入")

    elif Z > 100 or Z < 0:
        print("Z範圍只能0~100，請重新輸入")


    # 檢查限制條件
    elif X + 2*Y + 3*Z > 300:
        print("不符合限制條件：X+2Y+3Z 必須小於等於300")
        print("目前數值:", X + 2*Y + 3*Z)
        print("請重新輸入X,Y,Z")


    else:
        # 全部符合，跳出迴圈
        break



# 將輸入值組成PSO需要的格式
input_value = [X,Y,Z]


# 計算輸入組合的結果
result = -objective(input_value)


print("輸入的X,Y,Z:", input_value)
print("目標函數結果:", result)
```

```python
#結論:
#1.第2部分跟第十一部分都是用PSO求最小值,而第23部分用PSO求最大值並新增了手動輸入的部分
# 2. 第23部分改成 return -value 的目的：
# 因為 pyswarm 的 PSO 預設是進行最小化問題，
# 若希望尋找目標函數的最大值，需要將目標函數轉換為負值，
# 使 PSO 尋找 -value 的最小值，等同於找到 value 的最大值。
# 若寫 return value，則 PSO 會直接尋找 value 的最小值。

# 3. 第23部分改成 -fopt 與 result = -objective(input_value) 的原因：
# 因為前面將最大化問題轉換成最小化 -value，
# 所以 PSO 輸出的結果為負值，需要再乘以 -1 還原成原始目標函數的最大值。
```
